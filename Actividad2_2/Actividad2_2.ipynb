```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
```

```python
df= pd.read_csv("listings.csv")
df
```

```python
#5 host names diferentes
filtro1=df[(df["host_name"] == "Skyy") | (df["host_name"] == "Magdalena") | (df["host_name"] == "Laura") | (df["host_name"] == "Colby") | (df["host_name"] == "Edward")]
filtro1
```

```python
filtro1["host_name"].head(10)
```

```python
filtro1.to_csv("EjercicioA.csv")
```

```python
#5 hosts que se unieron después de 2020
filtro2=df[df["host_since"] >= "2021-01-01"]
filtro2
```

```python
filtro2["host_since"].head()
```

```python
filtro2.to_csv("EjercicioB.csv")
```

```python
#Responden a más tardar en 1 día
filtro3=df[(df["host_response_time"] == "within an hour") | (df["host_response_time"] == "within a few hours") | (df["host_response_time"] == "within a day")]
filtro3
```

```python
df["host_response_time"].head()
```

```python
filtro3.to_csv("EjercicioC.csv")
```

```python
#Multiplos de 200
filtro4=df.iloc[200::200, :]
filtro4
```

```python
filtro4.to_csv("EjercicioD.csv")
```

```python
#Entire home y private rooom
filtro5=df[(df["room_type"] == "Entire home/apt") | (df["room_type"] == "Private room")]
filtro5
```

```python
filtro5["room_type"]
```

```python
filtro5.to_csv("EjercicioE.csv")
```

```python
#Que tengan maximo 4 camas
filtro6=df[df["beds"] <= 4]
filtro6
```

```python
filtro6["beds"]
```

```python
filtro6.to_csv("EjercicioF.csv")
```

```python
#Es superhost
filtro7=df[df["host_is_superhost"] == "t"]
filtro7
```

```python
filtro7["host_is_superhost"]
```

```python
filtro7.to_csv("EjercicioG.csv")
```
